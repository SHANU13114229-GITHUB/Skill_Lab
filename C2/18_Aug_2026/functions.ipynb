```python
def find_avg_marks(marks):
    sum_of_marks = sum(marks)
    no_of_subjects = len(marks)
    avg_marks = sum_of_marks / no_of_subjects
    return avg_marks


def calc_grade(avg_marks):
    if avg_marks >= 80:
        return "A Grade"

    elif avg_marks >= 65:
        return "B Grade"

    elif avg_marks >= 55:
        return "C Grade"

    else:
        return "F Grade"


marks = list(map(int, input("Enter the Marks : ").split()))

avg_marks = find_avg_marks(marks)
print("Average marks:", avg_marks)
print("Grade:", calc_grade(avg_marks))
```

```python
def EvenOdd(num):
    if num%2==0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter a Number : "))
print(EvenOdd(num))
```
