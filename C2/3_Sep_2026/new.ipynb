```python
# Example: Using raise in Python

class NegativeSalaryError(Exception):
    """Custom exception for invalid salary values."""
    pass

def set_salary(amount):
    if amount < 0:
        # Raise a custom exception if salary is negative
        raise NegativeSalaryError("Salary cannot be negative!")
    return amount

try:
    # Valid case
    salary = set_salary(5000)
    print("Salary set to:", salary)

    # Invalid case
    salary = set_salary(-2000)
    print("Salary set to:", salary)

except NegativeSalaryError as e:
    print("Error:", e)

except Exception as e:
    # Catch any other unexpected errors
    print("Unexpected error:", e)
```
